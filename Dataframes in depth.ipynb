```python
# select all LB players
players[players.position =='LB']
```

```python
players[(players.position == "LB") & (players.age>=25)]
```

```python
# find out players who have market value 10+ and age is less than 25 and position is LB
players[(players.position == 'LB') & (players.age<=25) & (players.market_value >=10)]
```

```python
# find out players who have market value 10+ and age is less than 25 and position is LB not from Arsenal club or Tottenham club
players[(players.position == 'LB') & (players.age<=25) & (players.market_value >=10) & (players.club != 'Arsenal') & (players.club != 'Tottenham')]  # without using ~ operator
players[(players.position == 'LB') & (players.age<=25) & (players.market_value >=10) & ~(players.club.isin(['Arsenal','Tottenham']))]  # using ~ operator ~ means except given data
```
