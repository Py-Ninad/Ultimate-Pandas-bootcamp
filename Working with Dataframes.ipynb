#### Rename method

```python
name = [0,1,2,3]
age = [12.0,15.4,16.0,26.8]
height = [1.45,1.78,1.65,2.0]
df = pd.DataFrame({'Name':name,'Age':age,'Height':height},index=name)
```

```python
df
```
