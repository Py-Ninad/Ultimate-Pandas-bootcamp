```python
dataurl = 'https://andybek.com/pandas-nutrition'
```

```python
nutrition = pd.read_csv(dataurl)
```

```python
nutrition.sample(3)
# nutrition.head()
```

```python
nutrition.info()
```

```python
# dropping Unnamed:0 column as its extra
nutrition.drop('Unnamed: 0',axis=1)  # axis 1 stands for columns and axis 0 stands for rows
```

```python
nutrition.head()
```

```python
nutrition = pd.read_csv(dataurl, index_col=[0])  # here 0 is telling use column 0 as index while fetching data from dataurl
```

```python
nutrition
```

```python
nutrition.index
```

```python
nutrition = nutrition.set_index('name')
```

```python
nutrition.head(8)
```

```python
nutrition.loc["Eggplant, raw"]
```

```python
nutrition.loc["Cornstarch","calories"]  # df.loc[row,column]
```

```python
nutrition.loc["Sherbet, orange":"Lamb, raw, ground","calories":"sodium"]  # df.loc[[row:row],[column:column]] 
```

```python
nutrition.iloc[5,7]  # df.iloc[row,column]
```

```python
nutrition.iloc[[12,17,25],6:9]  # this is dataframe slicing iloc[[rows],columns slicing]
```

```python
nutrition.iloc[[10,20,30],[10,30,40]]  # this is dataframe slicing iloc[[rows],[columns]]
```
